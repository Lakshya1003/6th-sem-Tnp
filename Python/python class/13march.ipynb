```python
# fibnacci
def fibo(n):
  if(n == 0):
    return 0
  if(n == 1):
    return 1
  return fibo(n-2) + fibo(n-1)



a = 10
print(fibo(4))
```

```python
# factorial

def fac(n):
  if(n == 1):
    return 1
  return n*fac(n-1)



a = 5
print(fac(a))
```

```python
# sum of digits
def sd(val, n=0):
    if val == 0:
        return n
    else:
        n += val % 10
        return sd(val // 10, n)

a = 5
print(sd(a))
```

```python
# reverse a string
def rev(str1, i = 0):
  if(i >= len(str1)):
    return str1
  a = str1[i]
  b = str1[len(str1)-1-i]
  a,b = b,a
  # str1[i], = str1[len(str1)-1-i],str1[i]
  return rev(str1,i+1)


namee = "Jhon_cena"

print(rev(namee))
```

```python
# palindrome check

def pal(s , i = 0 , pali = False):
  if( i >= len(s)):
    pali = True
    return pali

  else:
    if( s[i] == s[len(s) - 1- i]):
      pali = True
      pal(s , i + 1 , pali)
      pali = False

    else:
      pali = False
      return pali



a = "naman"
print(pal(a))


```
